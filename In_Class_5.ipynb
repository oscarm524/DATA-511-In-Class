# Exercise 2 (5 points)

Define the term objective function in the context of linear programming and explain its role in optimization.

The **objective function** is a linear function representing the goal of the optimization, such as maximizing profit or minimizing cost. It is expressed in terms of decision variables and is subject to constraints. The linear programming process finds the values of these variables that optimize this function while satifying all given constraints.

# Exercise 4 (10 points)

A pastry shop produces two types of pastries: Croissants (X) and Danishes (Y). Each pastry requires:

| Resource | Croissant (X) | Danish (Y) | Available | 
| --- | --- | --- | --- |
| Flour (grams) | 200 | 100 | 5000 |
| Labor (minutes) | 10 | 15 | 400 |
| Profit per unit | $2 | $3 | |

The owner wants to maximize profit while not exceeding available resources. Formulate and solve this as a linear programming problem using Python.

```python
# We want to maximize 2X + 3Y
# Subject to the constraints:
# 200X + 100Y <= 5000 (flour constraint)
# 10X + 15Y <= 400 (labor constraint)
# X >= 0
# Y >= 0

from scipy.optimize import linprog

# Coefficients of the objective function (note: we maximize 2X + 3Y by minimizing -2X - 3Y)
c = [-2, -3]

# Coefficients of the inequality constraints
A = [
    [200, 100],  # flour constraint
    [10, 15]     # labor constraint
]
b = [5000, 400]

# Bounds for X and Y
x_bounds = (0, None)
y_bounds = (0, None)

# Solve the linear programming problem
result = linprog(c, A_ub=A, b_ub=b, bounds=[x_bounds, y_bounds])

# Display the results
if result.success:
    print("Optimal solution found:")
    print("X =", result.x[0])
    print("Y =", result.x[1])
    print("Maximum value of the objective function:", -result.fun)
else:
    print("No optimal solution found.")
```

# Exercise 5 (10 points)

Consider the context presented in Exercise 4. This time solve the program as an interger linear programming. Formulate and solve the problem as an integer programming problem in Python.

```python
import numpy as np
from scipy.optimize import milp, LinearConstraint, Bounds

# Objective function coefficients (maximize 2X + 3Y)
c = [-2, -3]

# Inequality constraints (flour and labor)
A = np.array([
    [200, 100],  # flour constraint
    [10, 15]     # labor constraint
])
b = np.array([5000, 400])

# Setting the constraints
linear_constraint = LinearConstraint(A, lb=-np.inf, ub=b)

# Setting the bounds for X and Y
bounds = Bounds([0, 0], [np.inf, np.inf])

# Integer constraints (if X and Y need to be integers)
integrality = np.array([1, 1])

# Solve the integer linear programming problem
result = milp(c, constraints=linear_constraint, bounds=bounds, integrality=integrality)

# Display the results
if result.success:
    print("Optimal solution found:")
    print("X =", result.x[0])
    print("Y =", result.x[1])
    print("Maximum value of the objective function:", -result.fun)
else:
    print("No optimal solution found.")
```
