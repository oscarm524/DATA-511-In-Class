# Exercise 5 (6 points)

A survey shows that 51.3% of those who like a new product are male. If 65% of respondents like the product, what is the probability that a randomly selected male likes the product (using Bayes’ Theorem)? Among those who don't like, 47.6% are male.

- P(Like) = 0.65
- P(Not Like) = 0.35
- P(Male|Like) = 0.513
- P(Male|Not Like) = 0.476

We want to find P(Like|Male).

P(Male) = P(Male and Like) + P(Male and Not Like)
        = P(Male|Like) x P(Like) + P(Male|Not Like) x P(Not Like)
        = 0.513 x 0.65 + 0.476 x 0.35
        = 0.5


P(Like | Male) = P(Like and Male) / P(Male)
               = P(Male|Like) x P(Like) / P(Male) 
               = 0.513 x 0.65 / 0.5
               = 0.667
