# Exercise 2 (5 points)

Compute the probability of getting exactly 2 heads in 3 tosses of a fair coin.

$X \sim Bin(n=3, p=0.5)$

We want to $P(X=2)$. Then,

$P(X=2) = \binom{3}{2}(0.5)^2(0.5)^{3-2} = 3 * 0.25 * 0.5 = 0.375$

```python
from scipy.stats import binom

prob = binom.pmf(k=2, n=3, p=0.5)
print(f"The probability of getting exactly 2 successes in 3 trials is {prob}")
```

# Exercise 4 (5 points)

For the standard normal distribution N(0, 1), what is the probability that a value falls between -1.3 and 1.7?

```python
from scipy.stats import norm

prob = norm.cdf(1.7) - norm.cdf(-1.3)
print(f"The probability of a standard normal variable being between -1.3 and 1.7 is {prob}")
```

# Exercise 5 (10 points)

A company manufactures two products: Product A and Product B. Each unit of Product A requires 2 hours of labor and 3 units of raw material. Each unit of Product B requires 4 hours of labor and 2 units of raw material. The company has a total of 100 labor hours and 90 units of raw material available. The profit for each unit sold is $40 for Product A and $50 for Product B. Find the optimal number of Products A and B that maximizes company's profit.

X: represents the number of Products A \
Y: represents the number of Products B

The goal is: Maximize the profit. The profit = 40X + 50Y (this is the objective function).

Labor constraint -> 2X + 4Y <= 100 \
material constraint -> 3X + 2Y <= 90

```python
from scipy.optimize import linprog

# Coefficients of the objective function (maximize the profit)
# Since linprog performs minimization, we use negative coefficients to maximize the profit
c = [-40, -50]

# Coefficcients for the constraints
# 2X + 4Y <= 100 (resource constraint)
# 3X + 2Y <= 90 (another resource constraint)
A = [
    [2, 4],
    [3, 2]
]
b = [100, 90]

# Bounds for the decision variables X and Y
x_bounds = (0, None)
y_bounds = (0, None)

# Solve the linear programming problem
res = linprog(c, A_ub=A, b_ub=b, bounds=[x_bounds, y_bounds])

if res.success:
    print(f"Optimal value of Product A is: {res.x[0]:.2f}")
    print(f"Optimal value of Product B is: {res.x[1]:.2f}")
    print(f"Maximum profit is: {-res.fun:.2f}")
else:
    print("No optimal solution found.")
    print(f"Solver message: {res.message}")
```
