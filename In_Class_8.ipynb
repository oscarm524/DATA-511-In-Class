## Exercise 3(a) (5 points)

Write a Python function to calculate the take rate from the notes.

```python
import numpy as np

def take_rate(price, beta0=3.6553, beta1=-0.55999):

    Z = beta0 + beta1 * price
    return 1 / (1 + np.exp(-Z))
```

## Exercise 3(c) (5 points)

Simulate contribution using: Revenue = Take Rate × Market Size × Price, and Cost = $1.50 per subscriber.

```python
def contribution(price, market_size=1e7, cost=1.5):

    tr = take_rate(price)
    revenue = tr * market_size * price
    total_cost = tr * market_size * cost
    return revenue - total_cost

# Simulate contribution fo a range of prices
prices = np.arange(0, 10, 0.01)
contributions = [contribution(p) for p in prices]

import matplotlib.pyplot as plt
plt.plot(prices, contributions, label="Contribution")
plt.xlabel('Price')
plt.ylabel('Contribution')
plt.axhline(0, color='black', linewidth=0.8, label="Break Even Point")
plt.title('Contribution vs Price')
plt.grid()
plt.legend()
plt.show();
```
